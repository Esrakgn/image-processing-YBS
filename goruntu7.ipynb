```python
import cv2
import numpy as np
```

```python
def cropped_black_borders(img:np.ndarray, thr =12, pad=0):
    #thr parametresi thresh değerini temsil ediyor. Arkaplan siyah renginde olduğu için 0'a çok yakın bir değer verdik.
    #pad parametresi padding'i temsil ediyor. Görüntüyü, arkaplandan ayırt ettikten sonra sol sağ üst ve alt taraflarında 
    #boşuk bırakıp bırakmamaya karar veriyor. Biz varsayılan 0 verdiğimiz için 4 taraftan da tamamen kesilmiş bir görüntü hedefliyoruz.

    #ilk olarak eğer görüntü renkli bir görüntüyse yani 3 kanallı ise bunun kontrolünü sağlıyoruz:

    if img.ndim==3:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    else:
        gray = img.copy()

    #görüntüyü tek kanallı hale çevirdikten sonra ilk olarak siyah pikselleri ayırt etmek için bir maske çıkarıyoruz

    mask = (gray>thr).astype(np.uint8)*255
    kernel = np.ones((5,5),dtype=np.uint8)

    #yukarıda elimizde bir harita (mask) ve bir matris var artık

    thresh = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel, iterations=2)

    # elimizdeki temiz maske görüntüsü içerisindem kontur listesini elde ediyoruz
    # kenerları birleştirip tek görüntü elde etmek amaç
    contour, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
   
    c = max(contour, key=cv2.contourArea)

    x,y,w,h = cv2.boundingRect(c)

    H , W = img.shape[:2]

    #görüntünün yükseklik ve genişliğini elde ettikten sonra cropped işlemine geçiyoruz
    #köşeleri tespit ediyoruz (4 köşe)

    x1 = max(0, x - pad)
    y1 = max(0, y - pad)
    x2 = min(W, x + w + pad)
    y2 = min(H, y + h+ pad)

    # taşma olmamasını sağladıktan sonra ilgili görüntüyü orj görüntüden koparıyoruz

    return img[y1:y2 , x1:x2]
                            
```

```python
my_bgr_img = cv2.imread("./yeni1.png")

my_cropped_img = cropped_black_borders(my_bgr_img)

cv2.imshow("Cropped ımage", my_cropped_img)
cv2.waitKey(0)
cv2.destroyAllWindows
```
