```python
height, width,channel = 150,210,3

my_black_img = np.zeros((height, width,channel),dtype= np.uint8)
my_white_img = np.ones((height, width,channel),dtype= np.uint8)*255
# iki farklı yoldan gri rengini oluşturalım
my_gray_img_1 = np.random.randint(127,128,size = (height, width,channel),dtype= np.uint8)
my_gray_img_2 = np.ones((height, width,channel),dtype= np.uint8)*127
```

```python
fig = plt.figure(figsize = (10,7),facecolor = "purple")
plt.subplot(2,2,1)
plt.imshow(my_black_img)
plt.axis("off")
plt.title("Black Image")

plt.subplot(2,2,2)
plt.imshow(my_white_img)
plt.axis("off")
plt.title("White Image")

plt.subplot(2,2,3)
plt.imshow(my_gray_img_1)
plt.axis("off")
plt.title("Gray Image 1")
                 
plt.subplot(2,2,4)
plt.imshow(my_gray_img_2)
plt.axis("off")
plt.title("Gray Image 2")
```

```python
# 3 kanallı RGB bir görüntüde kırmızı rengini [255,0,0]
# bu şu anlama gelir : red=255 green=0 blue=0
# yeşil renginin kodları = [0,255,0]
#mavi renginin kodları = [0,0,255]

#red green blue renklerini oluşturup tek bir plot'ta gösterin
height, width,channel = 150,210,3

my_red_img = np.zeros((height, width,channel),dtype= np.uint8)
my_green_img = np.zeros((height, width,channel),dtype= np.uint8)
my_blue_img = np.zeros((height, width,channel),dtype= np.uint8)

```

```python
my_red_img[:,:,0] = 255
my_green_img[:,:,1]= 255
my_blue_img[:,:,2] =255
```

```python
fig = plt.figure(figsize = (10,7),facecolor = "white")
plt.subplot(1,3,1)
plt.imshow(my_red_img)
plt.axis("off")
plt.title("Red Image")

plt.subplot(1,3,2)
plt.imshow(my_green_img)
plt.axis("off")
plt.title("Green Image")

plt.subplot(1,3,3)
plt.imshow(my_blue_img)
plt.axis("off")
plt.title("Blue Image 1")
```

```python
my_new_black_img= my_red_img * my_green_img * my_blue_img
my_new_white_img= my_red_img + my_green_img + my_blue_img
```

```python
# red = [255,0,0]
# green [0,255,0]
# blue = [0,0,255]
```

```python
fig = plt.figure(figsize = (10,7),facecolor = "pink")
plt.subplot(1,2,1)
plt.imshow(my_new_black_img)
plt.axis("off")
plt.title("Black Image")

plt.subplot(1,2,2)
plt.imshow(my_new_white_img)
plt.axis("off")
plt.title("White Image")

```

```python
# fenerbahçe renklerini çiziniz 
fener = np.zeros((height, width,channel),dtype=np.uint8)
```

```python
fener[:,:fener.shape[1]//2,:] = [255,255,0]
fener[:,fener.shape[1]//2:,:] = [0,0,255]
```

```python
fig = plt.figure(figsize = (10,7),facecolor = "white")

plt.imshow(fener)
plt.axis("off")
plt.title("Şampiyon Fenerbahçe")
```

```python
#color detection (renk tespit etme): bu işlem için görüntünün tüm satır ve sütununda ve/veya istediğimiz ya da tüm kanalları gezmek durumundayız

# örneğin fenerbahçe renklerini gs renklerinee dönüştürmek için önce fenerdeki mavi rengisonra bu kordinat sistemini çıkarmamız lazım
#sonra tekrar bir bütün indeks olarak kullanıp bunun değerini kırmızı olarak atamamız gerekir

#mavi = [0,0,255]
#0. indeks = 0 
#1. indeks =0 
#2. indeks =255
```

```python
koordinat_mavi = (fener[:,:,0]==0) & (fener[:,:,1]==0) &(fener[:,:,2]==255)
```

```python
#mavi renginin koordinatlarını tekrar fener amblemine yerleştiriyoruz, artık koordinat değilkoordinattaki değerler geliyor 
#şimdi mavi yerine kırmızının kodları 
# [255,0,0]

fener[koordinat_mavi] = [255,0,0]
```

```python
plt.imshow(fener)
plt.axis("off")
plt.title("En Şanslı takım")
```

```python
#gs amblemindeki sarı rengini beyaza , kurmızı rengini de siyaha dönüştürün 
#bu defa ikisini de tespit edip ikisini de değiştireceksiniz
```

```python

koordinant_sari = (fener[:,:,0]==255) & (fener[:,:,1]==255) & (fener[:,:,2]==0)
koordinant_kirmizi = (fener[:,:,0]==255) & (fener[:,:,1]==0) & (fener[:,:,2]==0)
```

```python
fener[koordinant_sari] = [255,255,255]
fener[koordinant_kirmizi] = [0,0,0]
```

```python
fig = plt.figure(figsize = (10,7),facecolor = "pink")
plt.imshow(fener)
plt.axis("off")
plt.title("yeni resim")
```
